```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB,BernoulliNB
from sklearn.metrics import(
accuracy_score,
precision_score,
recall_score,
f1_score,
)

```

```python
train_data =fetch_20newsgroups(
    subset="train",
    remove=("headers","footers","quotes")
)
```

```python
test_data=fetch_20newsgroups(
    subset="test",
    remove=("headers","footers","quotes")
)
```

```python
X_train_text=train_data.data
y_train=train_data.target
X_test_text=test_data.data
y_test=test_data.target
class_names=train_data.target_names
```

```python
print("Training documents:", len(X_train_text))
print("Testing documents :", len(X_test_text))
print("Number of classes :", len(class_names))
```

```python
vectorizer =CountVectorizer(
    stop_words="english",
    min_df=2
)
X_train=vectorizer.fit_transform(X_train_text)
X_test=vectorizer.transform(X_test_text)
```

```python
print("Training featurematrix:", X_train.shape)
print("Testing feature matrix :", X_test.shape)
```

```python
mnb= MultinomialNB()
mnb.fit(
    X_train,
    y_train
)
y_pred_mnb=mnb.predict(X_test)
```

```python
binary_vectorizer=CountVectorizer(
    stop_words="english",
    min_df=2,
    binary=True
)
X_train_binary=binary_vectorizer.fit_transform(
    X_train_text
)
X_test_binary=binary_vectorizer.transform(
    X_test_text
)
```

```python
bnb=BernoulliNB()
bnb.fit(
    X_train_binary,
    y_train
)
y_pred_bnb=bnb.predict(
    X_test_binary
)
```

```python
def evaluate(model_name, y_true, y_pred):
    print("\n",model_name,"\n")
    
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision :", precision_score(y_true, y_pred, average='weighted'))
    print("Recall :", recall_score(y_true, y_pred, average='weighted'))
    print("F1 Score :", f1_score(y_true, y_pred, average='weighted'))
```

```python
evaluate("Multinomial Naive Bayes", y_test, y_pred_mnb)
evaluate("Bernoulli Naive Bayes", y_test, y_pred_bnb)
```

```python
results=pd.DataFrame(columns=[
    "Model",
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
])

models=[
    ("MNB", mnb, y_pred_mnb),
    ("BNB", bnb, y_pred_bnb)
]

for name, model, pred in models:
    results.loc[len(results)]=[
        name,
        accuracy_score(y_test, pred),
        precision_score(y_test, pred,average='weighted'),
        recall_score(y_test, pred,average='weighted'),
        f1_score(y_test, pred,average='weighted'),
    ]
print(results)   
```

```python
results.set_index('Model', inplace=True)
results.plot(kind='bar', figsize=(10, 6))
plt.title('Model Performance Comparison')
plt.ylabel('Score')
plt.xticks(rotation=0)
```
